#2. 다양한 크기의 커널을 사용한 평균값 필터링

```python
from google.colab import drive
drive.mount('/content/gdrive', force_remount=True)

from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src = cv2.imread('/content/gdrive/My Drive/Colab Notebooks/data/rose.bmp', cv2.IMREAD_GRAYSCALE)

if src is None:
   print('Image load failed!')
   sys.exit()

cv2_imshow(src)

for ksize in (3, 5, 7):
    dst = cv2.blur(src, (ksize, ksize))
    desc = 'Mean: {}x{}'.format(ksize, ksize)
    cv2.putText(dst, desc, (10, 30), cv2.FONT_HERSHEY_SIMPLEX,
                1.0, 255, 1, cv2.LINE_AA)
    cv2_imshow(dst)
    # cv2.waitKey()

# cv2.destroyAllWindows()
```

#4. sigma 값을 이용하여 가우시안 필터링

```python
from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src= cv2.imread('/content/gdrive/My Drive/Colab Notebooks/data/rose.bmp', cv2.IMREAD_GRAYSCALE)
if src is None:
    print('Image load failed!')
    sys.exit()

cv2_imshow(src)

for sigma in range(1, 6):
# sigma 값을 이용하여 가우시안 필터링
    dst= cv2.GaussianBlur(src, (0, 0), sigma)

    desc = 'sigma = {}'.format(sigma)
    cv2.putText(dst, desc, (10, 30), cv2.FONT_HERSHEY_SIMPLEX,
    1.0, 255, 1, cv2.LINE_AA)

    cv2_imshow(dst)

# cv2.waitKey()
# cv2.destroyAllWindows()
```

#5. 언샤프 마스크 필터

```python
from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src= cv2.imread('/content/gdrive/My Drive/Colab Notebooks/data/rose.bmp', cv2.IMREAD_GRAYSCALE)

if src is None:
    print('Image load failed!')
    sys.exit()

blr= cv2.GaussianBlur(src, (0, 0), 2)
dst= np.clip(2.0*src-blr, 0, 255).astype(np.uint8)

cv2_imshow(src)
cv2_imshow(dst)

# cv2.waitKey()
# cv2.destroyAllWindows()
```

#6. 컬러 영상에 대한 언샤프 마스크 필터

```python
from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src= cv2.imread('/content/gdrive/My Drive/Colab Notebooks/data/rose.bmp')

if src is None:
    print('Image load failed!')
    sys.exit()

src_ycrcb= cv2.cvtColor(src, cv2.COLOR_BGR2YCrCb)

# 밝기 값만 가져와서 가우시안 필터링 적용
src_f= src_ycrcb[:, :, 0].astype(np.float32)
blr= cv2.GaussianBlur(src_f, (0, 0), 2.0)
src_ycrcb[:, :, 0] = np.clip(2. * src_f-blr, 0, 255).astype(np.uint8)

dst= cv2.cvtColor(src_ycrcb, cv2.COLOR_YCrCb2BGR)

cv2_imshow(src)
cv2_imshow(dst)

# cv2.waitKey()
# cv2.destroyAllWindows()
```
