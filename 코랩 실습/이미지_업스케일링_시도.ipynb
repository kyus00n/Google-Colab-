```python
from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src= cv2.imread('/content/gdrive/My Drive/Colab Notebooks/배경화면 후보/다이애나 in sea of Earth.jpg')

if src is None:
    print('Image load failed!')
    sys.exit()

src_ycrcb= cv2.cvtColor(src, cv2.COLOR_BGR2YCrCb)
ycrcb_planes= list(cv2.split(src_ycrcb)) # Convert tuple to list

# 밝기 성분에 대해서만 히스토그램 평활화 수행
ycrcb_planes[0] = cv2.equalizeHist(ycrcb_planes[0])
he_ycrcb = cv2.merge(ycrcb_planes) # Renamed to he_ycrcb for clarity: Histogram Equalized YCrCb

# 밝기 값만 가져와서 가우시안 필터링 적용 (히스토그램 평활화된 이미지에 적용)
he_ycrcb_planes = list(cv2.split(he_ycrcb)) # Split the histogram equalized image
src_f= he_ycrcb_planes[0].astype(np.float32) # Get the Y channel from the HE image

# 가우시안 필터 파라미터 및 샤프닝 강도 조절
sigma = 3.0  # @param {type:"number"} 가우시안 필터의 표준 편차 (높을수록 더 부드럽게 흐려짐)
alpha = 5.0  # @param {type:"number"} 샤프닝 강도 (높을수록 더 선명하게) - 이전 값보다 높여서 더 선명하게!

blr= cv2.GaussianBlur(src_f, (0, 0), sigma) # sigma 값 적용

# 언샤프 마스킹 공식: 선명한 이미지 = 원본 + alpha * (원본 - 블러된 이미지)
sharpened_y = np.clip(src_f + alpha * (src_f - blr), 0, 255).astype(np.uint8)

he_ycrcb_planes[0] = sharpened_y # 선명해진 Y 채널을 다시 병합

dst= cv2.cvtColor(cv2.merge(he_ycrcb_planes), cv2.COLOR_YCrCb2BGR) # Convert the modified HE image back to BGR

cv2_imshow(src)
cv2_imshow(dst)
```

```python
from google.colab.patches import cv2_imshow

import sys
import numpy as np
import cv2

src= cv2.imread('/content/gdrive/My Drive/Colab Notebooks/배경화면 후보/다이애나 in sea of Earth.jpg')

if src is None:
    print('Image load failed!')
    sys.exit()

# Convert to YCrCb color space
src_ycrcb = cv2.cvtColor(src, cv2.COLOR_BGR2YCrCb)

# Split channels
ycrcb_planes = list(cv2.split(src_ycrcb))

# Apply normalization to the Y (luminance) channel
# This scales the intensity values of the Y channel to the full 0-255 range
ycrcb_planes[0] = cv2.normalize(ycrcb_planes[0], None, 0, 255, cv2.NORM_MINMAX)

# Merge channels back
dst_ycrcb = cv2.merge(ycrcb_planes)

# Convert back to BGR
dst = cv2.cvtColor(dst_ycrcb, cv2.COLOR_YCrCb2BGR)

cv2_imshow(src)
cv2_imshow(dst)
```

### 이미지 해상도 높이기 (업스케일링)

이미지의 해상도를 높이는 것은 '업스케일링'이라고 합니다. 원본 이미지에 없는 새로운 세부 정보를 생성할 수는 없지만, 기존 픽셀을 기반으로 새로운 픽셀을 예측하여 이미지를 더 크게 만들고 시각적으로 부드럽게 보이게 할 수 있습니다.

OpenCV의 `cv2.resize` 함수를 사용하여 다양한 보간(interpolation) 방법을 통해 이미지 해상도를 높일 수 있습니다.

```python
# 원본 이미지 불러오기 (이미 위에서 로드된 src를 사용합니다.)
# src = cv2.imread('/content/gdrive/My Drive/Colab Notebooks/배경화면 후보/다이애나 in sea of Earth.jpg')

if src is None:
    print('Image load failed!')
    sys.exit()

# 해상도 높일 배율 설정 (예: 2배)
scale_factor = 2

# 새로운 해상도 계산
new_width = int(src.shape[1] * scale_factor)
new_height = int(src.shape[0] * scale_factor)
new_dimensions = (new_width, new_height)

# 다양한 보간법으로 이미지 리사이징 (업스케일링)
# cv2.INTER_NEAREST: 가장 빠르지만, 가장 낮은 품질 (블록처럼 보일 수 있음)
# cv2.INTER_LINEAR: 기본값, 비교적 빠르고 좋은 품질
# cv2.INTER_CUBIC: 4x4 픽셀 이웃 보간, 선명하지만 약간 느림
# cv2.INTER_LANCZOS4: 8x8 픽셀 이웃 보간, 가장 좋지만 가장 느림

resized_nearest = cv2.resize(src, new_dimensions, interpolation=cv2.INTER_NEAREST)
resized_linear = cv2.resize(src, new_dimensions, interpolation=cv2.INTER_LINEAR)
resized_cubic = cv2.resize(src, new_dimensions, interpolation=cv2.INTER_CUBIC)
resized_lanczos4 = cv2.resize(src, new_dimensions, interpolation=cv2.INTER_LANCZOS4)

print(f"Original Image Shape: {src.shape}")
print(f"Upscaled Image Shape ({scale_factor}x): {resized_linear.shape}")

cv2_imshow(src)
print('원본 이미지')

cv2_imshow(resized_nearest)
print('INTER_NEAREST 보간법 (블록처럼 보일 수 있음)')

cv2_imshow(resized_linear)
print('INTER_LINEAR 보간법 (기본값, 비교적 부드러움)')

cv2_imshow(resized_cubic)
print('INTER_CUBIC 보간법 (더 선명하지만 계산량 많음)')

cv2_imshow(resized_lanczos4)
print('INTER_LANCZOS4 보간법 (가장 높은 품질, 가장 계산량 많음)')
```
