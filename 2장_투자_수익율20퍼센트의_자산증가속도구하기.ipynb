```python
# 2장 # 종잣돈 1억원을 연평균 20% 투자 수익율로(워렌버핏의 투자 수익율)
# 불리면 자산은 어떻게 증가하는가?

import matplotlib.pyplot as plt

# 초기 투자 금액
initial_investment = 100000000  # 1억 원

# 연간 배당률
annual_dividend_rate = 0.20

# 투자 기간 (1년부터 30년까지)
years = list(range(1, 31))

# 매년 투자 금액 계산
investment_amounts = [
    initial_investment * (1 + annual_dividend_rate) ** year for year in years
]

# 그래프 생성
plt.figure(figsize=(10, 6))
plt.plot(years, investment_amounts, marker='o', linestyle='-', color='b', label='Investment Amount')

# 10년 단위로 금액 표기 (붉은 색)
for year in [10, 20, 30]:
    plt.text(year, investment_amounts[year - 1], f'{investment_amounts[year - 1]:,.0f} KRW',
             horizontalalignment='right' if year == 30 else 'center', color='red')

# 그래프 꾸미기
plt.title(f'30-Year Investment Growth with {annual_dividend_rate * 100}% Annual Dividend')
plt.xlabel('Investment Period (years)')
plt.ylabel('Amount (KRW)')
plt.xticks(years)
plt.legend()
plt.grid(True)
plt.show()
```
